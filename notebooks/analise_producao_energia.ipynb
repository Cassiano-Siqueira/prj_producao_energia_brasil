# 1. Introdução e Contextualização energética
O Brasil possui uma matriz energética diversificada. Este projeto analisa a evolução da produção de energia no Brasil entre os anos de 2015 e 2024, identificando as fontes predominantes, o crescimento das energias renováveis e a sazonalidade na geração de energia.

# 2. Explicação da base
A base de dados `simulacao_producao_energia_brasil.csv` contém informações temporais e geográficas sobre a geração e consumo de energia em MWh, capacidade instalada, emissão estimada de CO2 e percentual de fontes renováveis por estado e região.

# 3. Leitura dos dados

```python
import pandas as pd

df = pd.read_csv('../dados/simulacao_producao_energia_brasil.csv')
display(df.head())
```

# 4. Limpeza e preparação
Nesta etapa, verificamos os tipos de dados e valores nulos. A coluna 'data' será convertida para o formato datetime para permitir análises temporais corretas.

```python
print("Informações do DataFrame:")
df.info()
print("\nValores nulos por coluna:")
display(df.isnull().sum())

df['data'] = pd.to_datetime(df['data'])
print("\nTipo da coluna 'data' após conversão:", df['data'].dtype)
```

# 5. Engenharia de atributos e KPIs
Cálculo dos indicadores principais: Produção total, Fonte predominante e Estado com maior produção.

```python
producao_total = df['producao_mwh'].sum()
print(f"Produção Total do Brasil: {producao_total:,.2f} MWh")

producao_por_fonte = df.groupby('fonte_energia')['producao_mwh'].sum().sort_values(ascending=False)
fonte_predominante = producao_por_fonte.idxmax()
print(f"A fonte de energia predominante é: {fonte_predominante}")
display(producao_por_fonte)

producao_por_uf = df.groupby('uf')['producao_mwh'].sum().sort_values(ascending=False)
estado_lider = producao_por_uf.idxmax()
print(f"Estado que mais produz energia: {estado_lider}")
display(producao_por_uf.head(5))
```
